# Task 4: Statistical Hypothesis Testing

## Wine Quality Prediction and Statistical Analysis

**Objective:** Test relationships between wine chemical properties
and quality using inferential statistics.

**Library:** scipy.stats

**Significance level:** 0.05


```python

import pandas as pd
import numpy as np
from pathlib import Path
from scipy import stats

# Locate and load the cleaned dataset
possible_paths = [
    Path("../data/wine_quality_cleaned.csv"),
    Path("../wine_quality_cleaned.csv"),
    Path("data/wine_quality_cleaned.csv"),
    Path("wine_quality_cleaned.csv")
]

file_path = next((p for p in possible_paths if p.exists()), None)

if file_path is None:
    raise FileNotFoundError(
        "Cannot find wine_quality_cleaned.csv. Check the file location."
    )

df = pd.read_csv(file_path)

print("Dataset loaded successfully")
print("Dataset shape:", df.shape)
display(df.head())
```

```python

# Pearson correlation test
r, p_value = stats.pearsonr(df["alcohol"], df["quality"])

alpha = 0.05

print("PEARSON CORRELATION TEST")
print("Correlation coefficient:", round(r, 4))
print("Test statistic (r):", round(r, 4))
print("P-value:", p_value)
print("Significance level:", alpha)

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is statistically significant evidence")
    print("of a linear association between alcohol and quality.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is insufficient evidence of a linear association.")
```

```python

# Spearman correlation test
rho, p_spearman = stats.spearmanr(
    df["alcohol"],
    df["quality"]
)

print("SPEARMAN CORRELATION TEST")
print("Spearman correlation coefficient:", round(rho, 4))
print("Test statistic (rho):", round(rho, 4))
print("P-value:", p_spearman)
print("Significance level:", alpha)

if p_spearman < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is statistically significant evidence")
    print("of a monotonic association between alcohol and quality.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Insufficient evidence of a monotonic association.")
```

```python

# Create two quality groups
good_alcohol = df.loc[df["quality"] >= 7, "alcohol"]
not_good_alcohol = df.loc[df["quality"] < 7, "alcohol"]

print("Not Good group size:", len(not_good_alcohol))
print("Good group size:", len(good_alcohol))

print("\nMean alcohol content")
print("Not Good:", round(not_good_alcohol.mean(), 3))
print("Good:", round(good_alcohol.mean(), 3))

# Welch's independent t-test
t_stat, p_ttest = stats.ttest_ind(
    good_alcohol,
    not_good_alcohol,
    equal_var=False
)

print("\nWELCH'S T-TEST")
print("Test statistic:", round(t_stat, 4))
print("P-value:", p_ttest)
print("Significance level:", alpha)

if p_ttest < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The group mean alcohol contents differ significantly.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: Insufficient evidence of a difference in group means.")
```

```python

# Shapiro-Wilk normality test
shapiro_stat, shapiro_p = stats.shapiro(df["alcohol"])

print("SHAPIRO-WILK NORMALITY TEST")
print("Test statistic:", round(shapiro_stat, 4))
print("P-value:", shapiro_p)
print("Significance level:", alpha)

if shapiro_p < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The alcohol distribution shows evidence")
    print("of deviation from normality.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is insufficient evidence of deviation")
    print("from normality.")
```

```python

# 95% confidence interval for mean alcohol content
alcohol = df["alcohol"].dropna()

mean_alcohol = alcohol.mean()
standard_error = stats.sem(alcohol)
degrees_freedom = len(alcohol) - 1

lower, upper = stats.t.interval(
    confidence=0.95,
    df=degrees_freedom,
    loc=mean_alcohol,
    scale=standard_error
)

print("Mean alcohol content:", round(mean_alcohol, 4))
print("95% confidence interval:")
print("Lower bound:", round(lower, 4))
print("Upper bound:", round(upper, 4))
```

```python

# Statistical testing report

test_report = pd.DataFrame([
    {
        "Test": "Pearson correlation",
        "H0": "Population linear correlation = 0",
        "H1": "Population linear correlation != 0",
        "Alpha": alpha,
        "Test Statistic": r,
        "P-value": p_value,
        "Decision": "Reject H0" if p_value < alpha else "Fail to reject H0"
    },
    {
        "Test": "Spearman correlation",
        "H0": "Population rank correlation = 0",
        "H1": "Population rank correlation != 0",
        "Alpha": alpha,
        "Test Statistic": rho,
        "P-value": p_spearman,
        "Decision": "Reject H0" if p_spearman < alpha else "Fail to reject H0"
    },
    {
        "Test": "Welch independent t-test",
        "H0": "Population group means are equal",
        "H1": "Population group means differ",
        "Alpha": alpha,
        "Test Statistic": t_stat,
        "P-value": p_ttest,
        "Decision": "Reject H0" if p_ttest < alpha else "Fail to reject H0"
    }
])

display(test_report.round(5))

# Save report
test_report.to_csv("statistical_test_results.csv", index=False)

print("Statistical testing report saved.")
```

### Overall Statistical Conclusion

The Pearson and Spearman correlation tests identified statistically significant positive relationships between alcohol content and wine quality. Pearson's correlation coefficient was 0.4803, while Spearman's correlation coefficient was approximately 0.4880. Both tests produced p-values below the 0.05 significance level.

Welch's independent t-test showed a statistically significant difference in mean alcohol content between the two quality groups. Wines rated as good quality (quality ≥ 7) had a higher mean alcohol content (11.553) than wines rated as not good quality (quality < 7), with a t-statistic of 16.2095 and a p-value below 0.05.

The Shapiro–Wilk test indicated that alcohol content does not follow a normal distribution. Therefore, the distribution's non-normality should be considered when interpreting the statistical results.

Overall, the analysis suggests that alcohol content is associated with wine quality. However, correlation does not establish causation, and other chemical properties may also influence wine quality.

# Task 5: Statistical Modelling Using Statsmodels

## Objective
To build an Ordinary Least Squares (OLS) regression model to study the relationship between wine's chemical properties and its quality score.

The model will be evaluated using regression coefficients, p-values, confidence intervals, R-squared, residual diagnostics, and multicollinearity checks.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from pathlib import Path

# Load the cleaned dataset
file_path = Path("wine_quality_cleaned.csv")

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())
```

```python
# Select five relevant input variables
X = df[
    [
        "alcohol",
        "volatile acidity",
        "sulphates",
        "citric acid",
        "density"
    ]
]

# Target variable
y = df["quality"]

# Add an intercept (constant)
X = sm.add_constant(X)

# Fit the OLS regression model
ols_model = sm.OLS(y, X).fit()

# Display the model summary
print(ols_model.summary())
```

```python
# Extract coefficients, p-values, and confidence intervals
results = pd.DataFrame({
    "Coefficient": ols_model.params,
    "P-value": ols_model.pvalues,
    "CI Lower": ols_model.conf_int()[0],
    "CI Upper": ols_model.conf_int()[1]
})

results = results.round(4)

display(results)

# Identify statistically significant predictors
print("\nPredictors with p-values below 0.05:")
print(results[results["P-value"] < 0.05])
```

```python
# Extract overall model statistics
print("R-squared:", round(ols_model.rsquared, 4))
print("Adjusted R-squared:", round(ols_model.rsquared_adj, 4))
print("F-statistic:", round(ols_model.fvalue, 4))
print("F-statistic p-value:", ols_model.f_pvalue)

print("\nAIC:", round(ols_model.aic, 4))
print("BIC:", round(ols_model.bic, 4))
```

```python
# Get predicted values and residuals
predicted = ols_model.fittedvalues
residuals = ols_model.resid

print("Mean of residuals:", round(residuals.mean(), 6))
print("Residual standard deviation:", round(residuals.std(), 4))

# Plot 1: Residuals vs fitted values
plt.figure(figsize=(8, 5))
sns.scatterplot(x=predicted, y=residuals)
plt.axhline(y=0, color="red", linestyle="--")
plt.xlabel("Predicted Wine Quality")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.show()

# Plot 2: Residual distribution
plt.figure(figsize=(8, 5))
sns.histplot(residuals, kde=True)
plt.xlabel("Residuals")
plt.title("Distribution of OLS Residuals")
plt.show()

# Plot 3: Normal Q-Q plot
sm.qqplot(residuals, line="45", fit=True)
plt.title("Normal Q-Q Plot of Residuals")
plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Use predictors without the intercept
X_vif = df[
    [
        "alcohol",
        "volatile acidity",
        "sulphates",
        "citric acid",
        "density"
    ]
].copy()

# Calculate VIF for each predictor
vif_results = pd.DataFrame()
vif_results["Feature"] = X_vif.columns
vif_results["VIF"] = [
    variance_inflation_factor(X_vif.values, i)
    for i in range(X_vif.shape[1])
]

display(vif_results.round(3))
```

```python
# Save coefficients and confidence intervals
results.to_csv("ols_coefficients_report.csv", index=True)

# Save VIF results
vif_results.to_csv("ols_vif_report.csv", index=False)

# Save the complete OLS summary as a text file
with open("ols_model_summary.txt", "w") as file:
    file.write(ols_model.summary().as_text())

print("OLS coefficients report saved.")
print("VIF report saved.")
print("OLS model summary saved.")
```

# OLS Regression: Interpretation Report

## 1. Objective

The objective of this analysis is to examine the relationship between selected chemical properties and wine quality using Ordinary Least Squares (OLS) regression with the `statsmodels` library.

## 2. Variables Selected

Five input variables were selected:
- Alcohol
- Volatile acidity
- Sulphates
- Citric acid
- Density

The dependent variable is the wine quality score.

An intercept was included using `add_constant()` before fitting the OLS model.

## 3. Regression Coefficients

The regression coefficients describe the expected change in predicted wine quality associated with a one-unit increase in each predictor, while holding the other selected predictors constant.

A positive coefficient indicates a positive conditional association with wine quality, whereas a negative coefficient indicates a negative conditional association.

Predictor p-values were examined using a significance level of 0.05. The 95% confidence intervals were also reviewed to assess the uncertainty in each coefficient estimate.

## 4. Model Performance

R-squared measures the proportion of variation in wine quality explained by the selected predictors. Adjusted R-squared accounts for the number of predictors included in the model.

The F-statistic and its p-value assess whether the predictors collectively provide statistically significant explanatory power. A model p-value below 0.05 indicates that the model is statistically significant overall.

The numerical results are recorded in the OLS summary generated by `statsmodels`.

## 5. Residual Analysis

Residuals were calculated as the differences between observed and fitted wine quality scores. Residuals-versus-fitted plots, a residual histogram, and a normal Q-Q plot were examined.

These plots help assess whether the residuals show systematic patterns, unequal variance, or departures from approximate normality. The plots should be interpreted alongside the numerical results rather than relying on a single diagnostic.

## 6. Multicollinearity

Variance Inflation Factor (VIF) was calculated for each selected predictor. VIF values close to 1 indicate little multicollinearity, while larger values suggest that predictors may share overlapping information.

Predictors with high VIF values should be investigated before drawing strong conclusions about their individual coefficients.

## 7. Limitations

- The model uses only five chemical properties and may omit other relevant variables.
- OLS assumes a linear relationship between the predictors and the expected outcome.
- Wine quality is an integer rating, so a linear regression model may not fully represent its discrete nature.
- Outliers, heteroscedasticity, and non-normal residuals may affect statistical inference.
- Correlation and regression do not establish causation.
- Results from this dataset may not generalize to all wines or wine-producing regions.

## 8. Conclusion

OLS regression provides an interpretable way to investigate associations between chemical properties and wine quality. Regression coefficients, significance tests, confidence intervals, model-fit statistics, residual diagnostics, and VIF values should be considered together when evaluating the model.

The final conclusions should be based on the actual numerical results obtained from the fitted model, rather than assuming that every selected predictor is statistically significant.

# Task 6: Regression Model Development

## Objective
To predict numerical wine-quality scores using Linear Regression, Decision Tree Regressor, and Random Forest Regressor.

The models will be evaluated using MAE, MSE, RMSE, and R-squared. A fixed random seed will ensure reproducible train-test splits.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the cleaned dataset
file_path = Path("wine_quality_cleaned.csv")
df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:", df.isnull().sum().sum())
display(df.head())
```

```python
# Define the target variable
y = df["quality"]

# Select the input variables
X = df.drop(columns=["quality"])

print("Input features (X):")
print(X.columns.tolist())

print("\nNumber of input features:", X.shape[1])
print("Target variable:", y.name)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
# 1. Linear Regression
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

# 2. Decision Tree Regressor
tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X_train, y_train)

# 3. Random Forest Regressor
forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
forest_model.fit(X_train, y_train)

print("All three regression models trained successfully!")
```

```python
# Generate predictions using the three models

linear_predictions = linear_model.predict(X_test)

tree_predictions = tree_model.predict(X_test)

forest_predictions = forest_model.predict(X_test)

print("Predictions generated successfully!")

print("\nFirst 10 actual quality scores:")
print(y_test.head(10).values)

print("\nLinear Regression predictions:")
print(np.round(linear_predictions[:10], 2))

print("\nDecision Tree predictions:")
print(np.round(tree_predictions[:10], 2))

print("\nRandom Forest predictions:")
print(np.round(forest_predictions[:10], 2))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining class percentages:")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting class percentages:")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

print("Classification libraries imported successfully!")
```

```python
# Logistic Regression model

logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
# Decision Tree model

tree_classifier = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

tree_classifier.fit(X_train, y_train)

print("Decision Tree Classifier trained successfully!")
```

```python
# Random Forest model

forest_classifier = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

forest_classifier.fit(X_train, y_train)

print("Random Forest Classifier trained successfully!")
```

```python
# Generate predictions on test data

logistic_predictions = logistic_model.predict(X_test)

tree_predictions = tree_classifier.predict(X_test)

forest_predictions = forest_classifier.predict(X_test)

print("Predictions generated successfully!")

print("\nFirst 10 actual categories:")
print(y_test.head(10).values)

print("\nLogistic Regression predictions:")
print(logistic_predictions[:10])

print("\nDecision Tree predictions:")
print(tree_predictions[:10])

print("\nRandom Forest predictions:")
print(forest_predictions[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Function to evaluate classification models

def evaluate_classifier(actual, predicted):
    return {
        "Accuracy": accuracy_score(actual, predicted),
        "Precision": precision_score(
            actual, predicted, pos_label="Good", zero_division=0
        ),
        "Recall": recall_score(
            actual, predicted, pos_label="Good", zero_division=0
        ),
        "F1-Score": f1_score(
            actual, predicted, pos_label="Good", zero_division=0
        )
    }

# Evaluate all three models

classification_results = pd.DataFrame({
    "Logistic Regression": evaluate_classifier(
        y_test, logistic_predictions
    ),
    "Decision Tree": evaluate_classifier(
        y_test, tree_predictions
    ),
    "Random Forest": evaluate_classifier(
        y_test, forest_predictions
    )
}).T

display(classification_results.round(4))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

# Get probability of Good class
good_class_index = list(
    forest_classifier.classes_
).index("Good")

forest_probabilities = forest_classifier.predict_proba(X_test)[
    :, good_class_index
]

# Calculate ROC-AUC
roc_auc = roc_auc_score(
    (y_test == "Good").astype(int),
    forest_probabilities
)

print("Random Forest ROC-AUC:", round(roc_auc, 4))

# Plot ROC curve
fpr, tpr, thresholds = roc_curve(
    (y_test == "Good").astype(int),
    forest_probabilities
)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, label=f"Random Forest AUC = {roc_auc:.4f}")
plt.plot([0, 1], [0, 1], "r--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest")
plt.legend()
plt.show()
```
